```python
from ultralytics import YOLO

# Load baseline YOLOv8n
model = YOLO('yolov8n.pt')

# Train baseline with cache explicitly disabled to avoid missing file errors
results = model.train(
    data='../dataset.yaml',
    epochs=1,
    imgsz=640,
    batch=16,
    seed=42,
    patience=35,
    cache=False,  # <--- Disables reading/writing bad cache files
    name='underwater_refined_yolov8n'
)

print("Baseline training complete! Weights saved to: runs/detect/underwater_refined_yolov8n/weights/best.pt")
```
