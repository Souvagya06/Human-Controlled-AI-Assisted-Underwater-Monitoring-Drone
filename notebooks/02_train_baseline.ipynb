```python
from ultralytics import YOLO
import os

# Verify path configuration
yaml_path = "../dataset.yaml"
if os.path.exists(yaml_path):
    print("Found dataset.yaml successfully!")
else:
    print("Warning: dataset.yaml not found. Please check your path.")

# Load the lightweight YOLOv8 nano model
model = YOLO('yolov8n.pt')

# Start training on your refined enhanced dataset
results = model.train(
    data=yaml_path, 
    epochs=110, 
    imgsz=640, 
    batch=16, 
    name='underwater_refined_yolov8n'
)

print("Baseline training complete! Check your runs/detect/underwater_refined_yolov8n folder for metrics and results.")
```
