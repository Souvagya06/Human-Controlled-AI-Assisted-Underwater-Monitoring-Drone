```python
from ultralytics import YOLO
import os
import sys

# 1. Data sanity
yaml_path = "../dataset.yaml"
if os.path.exists(yaml_path):
    print("Found dataset.yaml successfully!")
else:
    print("ERROR: dataset.yaml not found. Please check your path.")
    sys.exit(1)

dataset_dir = os.path.dirname(yaml_path)
train_dir = os.path.join(dataset_dir, "images", "train")
val_dir = os.path.join(dataset_dir, "images", "val")
if not os.path.isdir(train_dir):
    print(f"ERROR: Training images directory not found: {train_dir}")
    sys.exit(1)
if not os.path.isdir(val_dir):
    print(f"ERROR: Validation images directory not found: {val_dir}")
    sys.exit(1)

# Load baseline YOLOv8n
model = YOLO('yolov8n.pt')

# Train baseline
results = model.train(
    data=yaml_path,
    epochs=110,
    imgsz=640,
    batch=16,
    seed=42,
    patience=35,
    name='underwater_refined_yolov8n'
)

print("Baseline training complete! Weights saved to: runs/detect/underwater_refined_yolov8n/weights/best.pt")
```
